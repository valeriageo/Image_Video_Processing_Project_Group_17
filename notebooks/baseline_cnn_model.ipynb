```python
# Environment Setup
from pathlib import Path
import sys
import torch

def find_project_root(start=None):
    start = (start or Path.cwd()).resolve()
    for candidate in [start, *start.parents]:
        if (candidate / "requirements.txt").exists() and (candidate / "src").exists():
            return candidate
    raise FileNotFoundError("Project root not found")

ROOT = find_project_root()
if str(ROOT) not in sys.path:
    sys.path.append(str(ROOT))

from src.baseline_cnn import BaselineCNN

# Instantiating the baseline CNN model with 10 output classes one for each digit.
model = BaselineCNN(num_classes=10)
print(model)

# Testing the model with a dummy input to verify the architecture and output form.
dummy = torch.randn(1, 1, 32, 32)
output = model(dummy)
print('Dummy output shape:', output.shape)
```
